```python
def even_or_odd(num):
    if num%2==0:
        print(f'{num} is even ')
    else:
        print(f'{num} is odd ')

even_or_odd(5)
even_or_odd(6)
even_or_odd(9)
even_or_odd(97)
even_or_odd(71)
even_or_odd(76)
even_or_odd(90)
even_or_odd(0)
even_or_odd(-3)
even_or_odd(5)
```
