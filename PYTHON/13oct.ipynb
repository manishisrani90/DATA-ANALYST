```python
# data serialization means converting a data into a other format and all
import json
details={"name":"manish","class":12,"age":21}
intojson=json.dumps(details)
print(intojson)
print(type(intojson))
prsdata=json.loads(intojson)
print(prsdata)
print(type(prsdata))
```

```python
# csv 
import csv
with open ('example.csv',mode='w',newline='') as file:
    write_the_file=csv.write_the_file(file)
    write_the_file.writerow(['name','age'])
    write_the_file.writerow(['krish',32])

with open('example.csv',mode='r') as file:
    read_the_file=csv.reader(file)
    for row in read_the_file:
        print(row)
```
