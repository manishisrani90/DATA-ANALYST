```python
def convertFtoC(temp,unit):
    if unit=='C':
        return temp*9/5 +32
    elif unit=='F':
        return(temp-32)*5/9
    else:
        print("null")


print( convertFtoC(33,'C'))

```

```python
# strong password checker
def is_strong_pass(password):
    if len(password)<8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char in '!@#$%^&*()_+' for char in password):
        return False
    return True



print(is_strong_pass("wK!jhnc1t"))
    
```

```python

def calculate_total_cost(cart):

    total_cost = 0

    for item in cart:
        total_cost += item['price'] * item['quantity']

    # Nested function
    def apply_discount(discount_percent):
        discount = total_cost * discount_percent / 100
        return total_cost - discount

    return apply_discount(10)


cart = [
    {'name': 'apple', 'price': 0.5, 'quantity': 4},
    {'name': 'Banana', 'price': 1.5, 'quantity': 12},
    {'name': 'kiwi', 'price': 2.5, 'quantity': 9},
    {'name': 'guava', 'price': 3.0, 'quantity': 19}
]

final_cost = calculate_total_cost(cart)
print(final_cost)
```

```python
def is_palindrome(s):
    s=s.lower().replace(" ","")
    return s==s[::-1]

print(is_palindrome("A man A PLAN A CANAL PANAMA"))
```
