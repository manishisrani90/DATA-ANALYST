```python
# vectorized operation using numpy
array_no_1=np.array([1,2,3,4,5,8])
array_no_2=np.array([1,1,5,9,5,8])

print("addition",array_no_1+array_no_2)
print(array_no_2-array_no_1)
print(np.sqrt(array_no_2))

print(np.sin(array_no_2))
print(np.log(array_no_2))
```
