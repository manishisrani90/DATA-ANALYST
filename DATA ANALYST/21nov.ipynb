```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame(
    {
        "Week": ["W1", "W2", "W3", "W4", "W5", "W6"],
        "Visitors": [420, 480, 460, 590, 640, 710],
    }
)
plt.plot(df["Week"],df["Visitors"],marker="o")
plt.title("Daily week and visitors trend ")
plt.xlabel("Week")
plt.ylabel("Visitor")
plt.show()
```

```python
import pandas as pd;
import matplotlib.pyplot as plt;

df=pd.read_csv("students.csv")
print(df.columns)
print(df.head())
plt.plot(df["age"], df["marks"], marker="o")
plt.title("Age vs Marks comparison ")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

```python
import matplotlib.pyplot as plt;
import pandas as pd;
df=pd.read_csv("students.csv")
city_counts=df["city"].value_counts()
print(city_counts)
plt.plot(city_counts.index,city_counts.values,color="red",marker="o",linestyle=":")
print(city_counts)
# plt.bar(city_counts.index, city_counts.values)
plt.title("Number of Students by City")
plt.xlabel("City")
plt.ylabel("Number of Students")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```
