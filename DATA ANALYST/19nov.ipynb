```python
x['marks_of_all']=x['marks'].fillna(x['marks'].mean())
```

```python
x=x.rename(columns={'attendance_pct':'Attendance'})
x=x.rename(columns={'marks_of_all':'Marks 2'})
x.head()
```

```python
grouped_mean=x.groupby('city')['Attendance'].mean()
print(grouped_mean)
```
